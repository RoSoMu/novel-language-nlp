# Comparative Analysis

This notebook brings together the findings from the preceding analyses of creative lexical forms in an adapted conversational corpus.

The investigation has examined three approaches to lexical representation:

- **Frequency-based representations:** Bag-of-Words and TF-IDF.
- **Static embeddings:** Pretrained word vectors from spaCy's `en_core_web_md` model.
- **Contextual embeddings:** Representations extracted from DistilBERT.

The objective is to compare what these approaches reveal about the lexical innovations *Frankenstein* and *frankensteining*, particularly their morphological forms, grammatical uses and discourse contexts.

The analysis draws on the linguistic annotations established during corpus inspection and the computational results obtained in the preceding notebooks.

Rather than treating similarity scores as direct measurements of meaning, we examine how the different representation methods support or constrain linguistic interpretation.

The comparison is exploratory and descriptive. It does not attempt to establish that a model understands the innovations or reconstructs their conversational development.

## 1. Comparison of representation methods

The preceding notebooks examined the conversational corpus using three approaches to lexical representation: frequency-based features, static word embeddings and contextual embeddings.

Although all three approaches produce numerical representations of linguistic data, they differ in their units of analysis, the information they encode and the comparisons they support.

The following table summarizes these differences in relation to the methods applied in this project.

| Feature | BoW / TF-IDF | Static embeddings (spaCy) | Contextual embeddings (DistilBERT) |
|---|---|---|---|
| **Representation unit** | Document–term features | Individual lexical tokens | Tokens or pooled subword representations within an input sequence |
| **Source of representation** | Frequencies calculated from the adapted corpus | Pretrained word vectors | Pretrained transformer representations conditioned on input context |
| **Context sensitivity** | No contextual representation of individual occurrences | Fixed vector for each lexical entry | Representation varies with surrounding tokens |
| **Novel lexical forms** | Can be included as vocabulary features if present in the corpus | May lack a pretrained vector | Can be represented through subword tokenization |
| **Main comparison in this project** | Lexical distributions across excerpts | Similarity between pretrained lexical vectors | Similarity between selected contextual occurrences |
| **Principal limitation** | Does not represent grammatical function or meaning directly | Cannot distinguish contextual uses of the same lexical entry | Similarity does not independently establish grammatical or semantic interpretation |

These differences are important when interpreting the results.

BoW and TF-IDF describe the distribution and relative weighting of lexical features within the adapted corpus. Static embeddings provide pretrained lexical representations but do not adapt them to individual occurrences. Contextual embeddings produce representations that vary with the input sentence or excerpt.

Consequently, the three approaches are not directly interchangeable, and their numerical outputs should not be compared as though they measured the same linguistic properties.

## 2. Comparative findings

The three representation approaches were applied to the same broader investigation of creative lexical forms, but they provided different kinds of evidence.

We compare their findings in relation to lexical coverage, morphological variation and contextual use, drawing on the results documented in the preceding notebooks.

### 2.1. Representation of creative lexical forms

The frequency-based analysis identified frankenstein and frankensteining as distinct lexical features in the adapted conversational corpus. Both forms were represented because they occurred in the analysed excerpts.

The static embedding analysis revealed a different pattern. Although spaCy tokenized both forms as individual tokens, only frankenstein had an available pretrained vector. The derived form frankensteining lacked a pretrained static representation in en_core_web_md.

DistilBERT, by contrast, represented both forms through its subword vocabulary:

* frankenstein → frankenstein
* frankensteining → frankenstein + ##ing

This allowed contextual representations to be extracted for both expressions, including the nominalised form that lacked a pretrained static vector in spaCy.

These findings illustrate an important distinction between lexical occurrence, pretrained vocabulary coverage and contextual representability.

A frequency-based method can include an unfamiliar expression as a feature when it occurs in the analysed corpus. A static embedding model may recognize that expression as a token without providing a pretrained vector. A subword-based transformer can construct a contextual representation from available token pieces.

However, the ability to represent an expression numerically does not establish that a model has identified its grammatical function or acquired its intended meaning.

### 2.2. Morphological and grammatical variation

The conversational corpus documents the development of *Frankenstein* from an innovative verbal use into subsequent nominal constructions involving *frankensteining*.

The linguistic annotations distinguish three relevant occurrences:

- **EX001:** *decided to Frankenstein it* — an innovative transitive verb occurring in an infinitival construction.
- **EX004:** *explaining the frankensteining and the unicorning* — a nominalised *-ing* form functioning as the head of a nominal expression.
- **EX010:** *Faculty of Counterfactual Frankensteining and Applied Gibberishonian Studies* — a nominalised *-ing* head modified by *Counterfactual*, occurring within the name of an imaginary academic institution.

These distinctions provide a linguistic basis for interpreting the computational representations.

**BoW and TF-IDF** distinguish the lexical forms *frankenstein* and *frankensteining* as separate features. However, the resulting document–term representations do not explicitly encode their grammatical functions or morphological relationship.

**Static embeddings** provide a pretrained vector for *frankenstein* but not for *frankensteining*. Consequently, the spaCy analysis cannot directly compare the two forms through their individual pretrained lexical vectors.

**Contextual embeddings** allow representations to be extracted for all three occurrences. DistilBERT tokenizes *frankenstein* as one subword token and *frankensteining* as two, enabling comparisons across the verbal and nominal constructions.

The contextual similarity results were:

| Comparison | Cosine similarity |
|---|---:|
| Constructed verbal vs. constructed nominal | 0.7008 |
| User verbal (EX001) vs. user nominal (EX004) | 0.5794 |
| Assistant nominal (EX010) vs. user verbal (EX001) | 0.5221 |
| Assistant nominal (EX010) vs. user nominal (EX004) | 0.7860 |
| Assistant nominal (EX010) vs. constructed nominal | 0.7633 |

The two highest cosine similarities occur between nominal expressions sharing the morphological form frankensteining. This pattern is compatible with an influence of shared morphology and grammatical function, although their individual contributions cannot be isolated.

The constructed verbal–nominal comparison (0.7008) also yields a higher similarity than either authentic comparison involving the original verbal occurrence (0.5794 and 0.5221). The constructed examples share more closely matched lexical and referential environments, including an explicit reference to the experimental dataset, which may contribute to their greater representational alignment.

Although the authentic occurrences belong to a shared conversational history, DistilBERT processes each excerpt independently. Their common discourse origin is therefore not directly available to the model, and the results do not establish whether this relationship contributes to their contextual representations.

### 2.3. Semantic reference and discourse context

The linguistic annotations document not only the grammatical functions of the target expressions but also their semantic references and discourse functions.

In EX001, the innovative verb *Frankenstein* refers to the reconstruction and combination of material from the original psycholinguistic experiment.

In EX004, the nominalised form *frankensteining* refers to the resulting process, now treated as an identifiable activity within the conversation.

In EX010, the assistant extends the expression into an imaginary academic context, incorporating *Frankensteining* into the name of a fictional faculty.

These occurrences are connected through the conversational development of the lexical innovation. However, that relationship is not explicitly represented in the isolated excerpts supplied to DistilBERT.

The contextual embeddings therefore reflect the linguistic environments available within each input, rather than the complete history through which the expressions acquired their intended references.

This distinction is particularly relevant to the observed similarity pattern.

The two highest similarities occur between nominal expressions sharing the form *frankensteining* (0.7860 and 0.7633), despite differences in their immediate contexts and discourse references.

Meanwhile, the constructed verbal–nominal pair (0.7008) exhibits greater similarity than the authentic comparisons involving the original verbal occurrence (0.5794 and 0.5221). The constructed examples share more closely matched lexical and referential environments, including an explicit reference to the experimental dataset.

These observations suggest that shared morphological form, grammatical function and local lexical context are plausible contributors to the observed similarities. Their individual effects cannot be isolated from the present comparisons.

The shared conversational origin of the authentic expressions does not, by itself, produce higher similarity scores than those observed between the constructed examples.

Consequently, the results raise a further research question: **how do grammatical function, morphological form and surrounding lexical context contribute to contextual similarity when these factors are systematically controlled?**

Addressing this question would require a separate experiment with additional examples and controlled linguistic contrasts.

## 3. Methodological limitations

The comparative analysis is exploratory and descriptive. Its findings should be interpreted in relation to the characteristics of the corpus, the representation methods and the design of the comparisons.

### 3.1. Corpus size and selection

The analysis uses a small, manually curated set of conversational excerpts containing creative lexical forms.

The corpus was adapted from a longer interaction, and the selected excerpts do not preserve the complete conversational history through which the expressions developed their meanings and discourse functions.

Consequently, the findings describe the behaviour of the selected representation methods on these examples. They cannot be generalized to creative lexical innovations more broadly without additional data.

### 3.2. Pretrained representations

Both the spaCy and DistilBERT analyses use pretrained models without additional training or fine-tuning on the conversational corpus.

The static embeddings reflect representations learned from external training data, while the contextual embeddings are conditioned on the linguistic material supplied in each input.

Neither model was explicitly trained to recognise the locally established meanings of *Frankenstein* and *frankensteining* within this conversation.

The availability of numerical representations therefore does not establish that the models have acquired those meanings.

### 3.3. Confounding linguistic variables

The contextual comparisons involve expressions that differ simultaneously in several linguistic properties, including morphological form, grammatical function, lexical surroundings and semantic reference.

In particular, the verbal examples contain *Frankenstein*, whereas the nominal examples contain *frankensteining*. Morphological form and grammatical function are therefore not independently controlled.

The constructed examples also differ from the authentic excerpts in lexical overlap, referential content and sentence structure.

As a result, the observed cosine similarities cannot be attributed exclusively to any individual linguistic factor.

### 3.4. Interpretation of similarity

Cosine similarity measures directional alignment between numerical representations. It does not directly measure semantic understanding, grammatical recognition or the ability to reconstruct conversational history.

The comparisons involve a small number of selected embeddings, and several pairwise comparisons reuse the same occurrences. They should therefore not be treated as independent observations supporting statistical generalization.

Furthermore, the similarity values obtained from different representation methods are not directly comparable as measurements of a common linguistic property.

The results are best understood as descriptive evidence of how the selected methods represent the target expressions under the conditions examined.

### 3.5. Conversational context

The authentic occurrences belong to an interaction in which the intended meanings of the lexical innovations were progressively established and extended.

However, DistilBERT processed each selected excerpt independently, without access to the preceding exchanges.

The model therefore received the immediate linguistic context of each expression but not the full discourse history available to the participants.

This distinction limits what can be inferred about the relationship between contextual representations and meaning developed through interaction.

A future investigation could examine longer conversational sequences and systematically controlled linguistic contrasts to explore these relationships further.

## 4. Conclusions and future research

### 4.1. Conclusions

This project investigated how three NLP representation approaches handle creative lexical forms and their contextual uses in an adapted conversational corpus.

The comparison revealed important differences between the methods.

Bag-of-Words and TF-IDF represented the target expressions as distinct lexical features, allowing their distributions to be examined across the selected excerpts. However, these methods did not explicitly represent their morphological relationship or grammatical functions.

The pretrained static embedding model provided a vector for *frankenstein* but not for the derived form *frankensteining*. Its lexical similarity analysis also revealed unexpected numerical identities between vectors assigned to apparently unrelated vocabulary entries, illustrating the limitations of interpreting static vector proximity as semantic equivalence.

DistilBERT provided contextual representations for both expressions through subword tokenization. The resulting similarities varied across the selected verbal and nominal constructions.

The highest similarities occurred between nominal expressions sharing the form *frankensteining*, while the constructed verbal–nominal comparison showed greater similarity than the authentic comparisons involving the original verbal occurrence.

These patterns are compatible with contributions from shared morphology, grammatical function and surrounding lexical context. However, the present comparisons do not isolate these factors or establish their individual effects.

The linguistic annotations were essential for interpreting the computational results. They documented grammatical functions, semantic references and discourse relationships that could not be inferred reliably from the numerical representations alone.

Overall, the findings demonstrate the value of combining linguistic analysis with computational representations when investigating creative lexical forms.

They also highlight the importance of distinguishing **the ability to produce a numerical representation from the ability to identify a linguistic function or recover an intended meaning**.

### 4.2. Future research

The findings suggest several directions for further investigation.

**Automatic linguistic annotation.** POS tagging and named entity recognition could be applied to the corpus to examine how pretrained models classify innovative verbal forms, nominalisations and noun-premodifying constructions. This could include *unicorning*, *Gibberishonian* and the fictional Faculty of Counterfactual Frankensteining and Applied Gibberishonian Studies.

**Controlled contextual comparisons.** Additional constructed examples could systematically vary morphological form, grammatical function and surrounding lexical context. This would allow the relative contributions of these factors to contextual similarity to be investigated more directly.

**Expanded conversational context.** A larger collection of exchanges could preserve more of the interaction in which the lexical innovations emerged and developed. This would support further investigation of the relationship between local linguistic context and meaning established through discourse.

These directions are proposed as potential extensions rather than conclusions established by the present analysis.

The annotated corpus and accompanying notebooks provide a documented starting point for further exploration of creative lexical morphology and its computational representation.
